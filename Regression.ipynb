```python
import numpy as np
import matplotlib.pyplot as plt
import math
```

```python
x_train=np.array([1.0,2.0])
y_train=np.array([300.0,500.0])

m=x_train.shape[0] # number of training examples
i=0
for i in range (m):
    x_i=x_train[i]
    y_i=y_train[i]
    print(f"x^{i}, y^{i} = ({x_i}, {y_i})")
```

```python
# Plotting actual training data
plt.scatter(x_train,y_train,marker='x',c='b')
plt.title("Training Data")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
w=100
b=100
```

```python
def compute_model_function(x,w,b):
    m=x.shape[0]
    f_wb=np.zeros(m)
    for i in range(m):
        f_wb[i]=w*x[i]+b
    return f_wb

# this function finds the predicted values based on the values of w and b
```

```python
tmp_f_wb=compute_model_function(x_train,w,b)
plt.scatter(x_train, y_train, marker='x', c='b')
plt.plot(x_train, tmp_f_wb, c='r')
plt.title("Title")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
def compute_cost(x,y,w,b):
    m=x.shape[0]
    cost_sum=0
    for i in range(m):
        f_wb=w*x[i]+b
        cost=(f_wb-y[i])**2
        cost_sum=cost_sum+cost
    total_cost=(cost_sum)/(2*m)
    return total_cost
# this function computes how bad the predicted value is ( counts the error )
```

```python
def compute_gradient(x,y,w,b):
    m=x.shape[0]
    dj_dw=0
    dj_db=0
    for i in range (m):
        f_wb=w*x[i]+b
        dj_dw_i=(f_wb-y[i])*x[i]
        dj_db_i=(f_wb-y[i])
        dj_db+=dj_db_i
        dj_dw+=dj_dw_i
    dj_dw/=m
    dj_db/=m
    return dj_dw,dj_db
# calculates the gradient of the cost function at the current values of w and b
```

```python
def plot_gradients(x,y,cost,grad):

    fig, ax = plt.subplots(1, 2, figsize=(12, 5))

    b_fixed = 100

    w_array = np.linspace(0, 400, 200)
    cost_array = [compute_cost(x_train, y_train, w, b_fixed)
                  for w in w_array]

    ax[0].plot(w_array, cost_array)
    ax[0].set_title("Cost vs w, with gradient; b set to 100")
    ax[0].set_xlabel("w")
    ax[0].set_ylabel("Cost")

    sample_ws = [100, 200, 300]

    for w in sample_ws:

        cost = compute_cost(x_train, y_train, w, b_fixed)
        dj_dw, dj_db = compute_gradient(
            x_train, y_train, w, b_fixed)

        ax[0].scatter(w, cost)

        # tangent line
        x_line = np.linspace(w-40, w+40, 2)
        y_line = cost + dj_dw*(x_line-w)

        ax[0].plot(x_line, y_line, 'r--')

        ax[0].annotate(
            rf'$\frac{{\partial J}}{{\partial w}}={dj_dw:.0f}$',
            (w+10, cost+1000)
        )

    w_vals = np.linspace(-100, 600, 10)
    b_vals = np.linspace(-200, 200, 10)

    W, B = np.meshgrid(w_vals, b_vals)

    U = np.zeros_like(W)
    V = np.zeros_like(B)

    for i in range(W.shape[0]):
        for j in range(W.shape[1]):

            dj_dw, dj_db = compute_gradient(
                x_train,
                y_train,
                W[i, j],
                B[i, j]
            )

            U[i, j] = -dj_dw
            V[i, j] = -dj_db

    ax[1].quiver(W, B, U, V)

    ax[1].set_title("Gradient shown in quiver plot")
    ax[1].set_xlabel("w")
    ax[1].set_ylabel("b")

    plt.tight_layout()
    plt.show()
```

```python
plot_gradients(x_train,y_train,compute_cost,compute_gradient)
```

```python
def gradient_descent(x,y,w_in,b_in,alpha,num_iter,cost_fn,gradient_fn):
    J_history=[]
    p_history=[]
    b=b_in
    w=w_in
    for i in range(num_iter):
        dj_dw,dj_db=gradient_fn(x,y,w,b)
        b=b-alpha*dj_db
        w=w-alpha*dj_dw
        if i<100000:
            J_history.append(cost_fn(x,y,w,b))
            p_history.append([w,b])
        if i%math.ceil(num_iter/10)==0:
            print(
                f"Iteration {i:4}: "
                f"Cost {J_history[-1]:0.2e} "
                f"dj_dw: {dj_dw:0.3e} "
                f"dj_db: {dj_db:0.3e} "
                f"w: {w:0.3e}, "
                f"b: {b:0.3e}"
            )
        
    return w,b,J_history,p_history
```

```python

w_init = 0
b_init = 0

# some random values
iterations = 10000
tmp_alpha = 1.0e-2


w_final, b_final, J_hist, p_hist = gradient_descent(x_train ,y_train, w_init, b_init, tmp_alpha, 
                                                    iterations, compute_cost, compute_gradient)
print(f"(w,b) found by gradient descent: ({w_final:8.4f},{b_final:8.4f})")
```

```python
# plotting cost versus iteration  

fig, (ax1, ax2) = plt.subplots(1, 2, constrained_layout=True, figsize=(12,4))
ax1.plot(J_hist[:100])
ax2.plot(1000 + np.arange(len(J_hist[1000:])), J_hist[1000:])
ax1.set_title("Cost vs. iteration(start)");  ax2.set_title("Cost vs. iteration (end)")
ax1.set_ylabel('Cost')            ;  ax2.set_ylabel('Cost') 
ax1.set_xlabel('iteration step')  ;  ax2.set_xlabel('iteration step') 
plt.show()
```

```python
def plt_contour(x,y,p_hist,ax,
                w_range=(-100, 500),
                b_range=(-500, 500),
                contours=(1000, 2000, 5000, 10000, 25000, 50000),
                resolution=1):

    # Create the grid
    w_vals = np.arange(w_range[0],w_range[1] + resolution,resolution)

    b_vals = np.arange(b_range[0],b_range[1] + resolution,resolution)

    W, B = np.meshgrid(w_vals, b_vals)

    # Compute the cost surface
    J = np.zeros_like(W, dtype=float)

    for i in range(W.shape[0]):
        for j in range(W.shape[1]):
            J[i, j] = compute_cost(x,y,W[i, j],B[i, j])

    # Draw the contours
    cs = ax.contour(W,B,J,levels=contours)

    ax.clabel(cs, inline=True, fontsize=8)

    # draw the gradient descent path
    w_hist = [p[0] for p in p_hist]
    b_hist = [p[1] for p in p_hist]

    ax.plot(
        w_hist,
        b_hist,
        'r.-',
        linewidth=2
    )

    # draw the arrows
    step = max(1, len(p_hist)//30)

    for i in range(step, len(p_hist), step):
        ax.arrow(
            w_hist[i-step],
            b_hist[i-step],
            w_hist[i]-w_hist[i-step],
            b_hist[i]-b_hist[i-step],
            color='r',
            head_width=0.001,
            head_length=0.0002,
            length_includes_head=True
        )
    ax.set_xlim(w_range[0], w_range[1])
    ax.set_ylim(b_range[0], b_range[1])
    ax.set_xlabel("w")
    ax.set_ylabel("b")
    ax.set_title(
        "Contour plot of cost J(w,b), with path of gradient descent"
    )
```

```python
fig, ax = plt.subplots(1, 1, figsize=(12, 6))
plt_contour(x_train,y_train,p_hist,ax)
plt.show()
```

```python
def plot_divergence(p_hist, J_hist, x_train, y_train):

    from mpl_toolkits.mplot3d import Axes3D

    fig = plt.figure(figsize=(12,4))

    ax1 = fig.add_subplot(121)
    w_array = np.linspace(-70000,70000,200)
    cost_array = [compute_cost(x_train,y_train,w,100) for w in w_array]
    ax1.plot(w_array,cost_array)
    w_hist = [p[0] for p in p_hist[:10]]
    ax1.plot(w_hist,J_hist[:10],color='magenta',linewidth=3)
    ax1.set_title("Cost vs w, b set to 100")
    ax1.set_xlabel("w")
    ax1.set_ylabel("Cost")


    ax2 = fig.add_subplot(122, projection='3d')
    w_vals = np.linspace(-70000,70000,80)
    b_vals = np.linspace(-40000,30000,80)
    W,B = np.meshgrid(w_vals,b_vals)
    J = np.zeros_like(W)
    for i in range(W.shape[0]):
        for j in range(W.shape[1]):
            J[i,j] = compute_cost(x_train,y_train,W[i,j],B[i,j])

    ax2.plot_surface(W,B,J,alpha=0.4,cmap='Blues')
    w_hist = [p[0] for p in p_hist[:10]]
    b_hist = [p[1] for p in p_hist[:10]]
    J_path = [compute_cost(x_train,y_train,w,b) for w,b in p_hist[:10]]
    ax2.plot(w_hist,b_hist,J_path,color='magenta',linewidth=3)
    ax2.set_xticks([-60000,-30000,0,30000, 60000])
    ax2.set_yticks([-60000,-30000,0,30000, 60000])
    ax2.set_xlabel("w")
    ax2.set_ylabel("b")
    ax2.set_zlabel("cost")
    ax2.view_init(elev=25, azim=-60)


    plt.tight_layout()
```

```python
plot_divergence(p_hist,J_hist,x_train,y_train)
plt.show()
```
